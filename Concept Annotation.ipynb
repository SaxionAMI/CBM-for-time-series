## Concept Annotation

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
class_names = [
    "Standing",       # L1
    "Sitting",        # L2
    "Lying",          # L3
    "Walking",        # L4
    "Stairs",         # L5
    "WaistBends",     # L6
    "ArmRaise",       # L7
    "KneeBend",       # L8
    "Cycling",        # L9
    "Jogging",        # L10
    "Running",        # L11
    "Jump"            # L12
]
```

### Set concept name 

```python
# snake_case
concept_name = "body_elevation"
```

### Select labels and concept value

```python
selected = [ ""

# , "Standing"       # L1
# , "Sitting"        # L2
# , "Lying"          # L3
# , "Walking"        # L4
# , "Stairs"         # L5
# , "WaistBends"     # L6
# , "ArmRaise"       # L7
# , "KneeBend"       # L8
# , "Cycling"        # L9
# , "Jogging"        # L10
# , "Running"        # L11
, "Jump"           # L12
            
]

concept_value = 2

selected = selected[1:]
print(f'Labels: {selected}')
print(f'Value: {concept_value}')
```

### Generate/update concept CSVs

```python
print(f'Concept: {concept_name}')
for i in range(1, 11):
    print(f"Subject {i}")
    subject_name = f"subject{i}"
    
    subject = pd.read_csv(f'dataset/mHealth_subject{i}.log', sep='\s+', header=None)
    label_series = subject[23]
    csv_length = len(label_series)
    
    csv_file = f"concepts/{concept_name}/{concept_name}_{subject_name}.csv"
    os.makedirs(os.path.dirname(csv_file), exist_ok=True)

    if os.path.exists(csv_file):
        df = pd.read_csv(csv_file)
    else:
        # initialize with nan values so the column exists from the start
        df = pd.DataFrame({"value": [float('nan')] * csv_length})

    for label in selected:
        print(f"Updating label with value {concept_value} : {label}")
        label_index = class_names.index(label) + 1
        
        mask = (label_series == label_index)
        df.loc[mask, "value"] = concept_value
        
    df.to_csv(csv_file, index=False)
```
